# Lab 1: Your First Agent

# Lab 2: Add Tools

https://learn.microsoft.com/en-us/agent-framework/get-started/add-tools?pivots=programming-language-python



```python
from agent_framework import Agent
from agent_framework.openai import OpenAIChatClient

# 사용할 모델명
MODEL = "gpt-5.6-luna"

# 지정한 모델로 OpenAI 채팅 클라이언트를 생성한다.
client = OpenAIChatClient(model=MODEL)
```

```python
from agent_framework import tool
from typing import Annotated
from random import randint
from pydantic import Field
from agent_framework import tool

# 참고: approval_mode="never_require"는 예제를 간결하게 하기 위한 설정이다.
# 실제 운영 환경에서는 도구 실행 전 사용자 확인을 받도록 "always_require"를 사용한다.
@tool(approval_mode="never_require")
def get_weather(
    location: Annotated[str, Field(description="날씨를 조회할 지역.")],
) -> str:
    """주어진 지역의 날씨를 가져온다."""
    # 무작위로 선택할 날씨 상태 목록
    conditions = ["맑음", "흐림", "비", "폭풍"]
    # 날씨 상태와 최고 기온을 무작위로 골라 문장으로 만들어 반환한다.
    return f"{location}의 날씨는 {conditions[randint(0, 3)]}이고 최고 기온은 {randint(10, 30)}°C입니다."
```

```python
# 이름, 지시문, 도구 목록을 지정해 날씨 에이전트를 생성한다.
agent = Agent(
    client=client,
    name="WeatherAgent",
    instructions="너는 도움이 되는 날씨 에이전트다. 질문에 답할 때 get_weather 도구를 사용하라.",
    tools=[get_weather],
)
```

```python
# <run_agent>
# 에이전트에 입력 메시지를 전달해 실행하고 결과를 받는다.
result = await agent.run("시애틀 날씨는 어때?")

# 결과를 출력한다.
print(f"에이전트: {result}")
```

# Lab3  Multi-Turn Conversations



https://learn.microsoft.com/en-us/agent-framework/get-started/multi-turn?pivots=programming-language-python


```python
from agent_framework import Agent
from agent_framework.openai import OpenAIChatClient

# 사용할 모델명
MODEL = "gpt-5.6-luna"

# 지정한 모델로 OpenAI 채팅 클라이언트를 생성한다.
client = OpenAIChatClient(model=MODEL)
```

```python
# 이름과 지시문을 지정해 대화용 에이전트를 생성한다.
agent = Agent(
    client=client,
    name="ConversationAgent",
    instructions="너는 친근한 어시스턴트다. 답변은 간결하게 유지하라.",
)
```

```python
# 대화 기록을 유지하기 위한 세션을 생성한다.
session = agent.create_session()

# 첫 번째 턴
result = await agent.run("내 이름은 앨리스고 등산을 좋아해.", session=session)
print(f"에이전트: {result}\n")
```

```python

# 두 번째 턴 — 에이전트가 사용자의 이름과 취미를 기억해야 한다.
result = await agent.run("나에 대해 뭘 기억하고 있어?", session=session)
print(f"에이전트: {result}")
```
